**A. Membangun Katalog (Create)** *(bobot 20%)*

Buat database baru bernama `kopi_nusantara`, collection bernama `menu`. Sisipkan **minimal 12 dokumen** menu dengan field bervariasi sesuai kategorinya:
- Menu kategori **"Kopi"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `tingkat_kepahitan` (skala 1-5).
- Menu kategori **"Makanan"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `pedas_level` (skala 0-3).
- Boleh menambahkan field lain sesuai kreativitas anda (mis. `stok`, `rekomendasi_barista`).

```python
from pymongo import MongoClient

# Membuat koneksi ke MongoDB yang berjalan di localhost, port default 27017
client = MongoClient("mongodb://localhost:27017/")

# (1) Membuat database dan collection baru
db = client["kopi_nusantara"]
koleksi_menu = db["menu"]

# (2) Menyisipkan banyak dokumen sekaligus
koleksi_menu.delete_many({})
koleksi_menu.insert_many([
    {"nama_menu": "Americano", "harga": 15000, "kategori": "Kopi" , "tingkat_kepahitan": 5},
    {"nama_menu": "Avocado Coffee", "harga": 28000, "kategori": "Kopi" , "tingkat_kepahitan": 2},
    {"nama_menu": "Latte", "harga": 22000, "kategori": "Kopi" , "tingkat_kepahitan": 3},
    {"nama_menu": "Caramel Latte", "harga": 26000, "kategori": "Kopi" , "tingkat_kepahitan": 1},
    {"nama_menu": "Cappuccino", "harga": 22000, "kategori": "Kopi" , "tingkat_kepahitan": 4},
    {"nama_menu": "Caramel Macchiato", "harga": 28000, "kategori": "Kopi" , "tingkat_kepahitan": 2},
    {"nama_menu": "Kopi Dolce", "harga": 22000, "kategori": "Kopi" , "tingkat_kepahitan": 2},
    {"nama_menu": "Seblak", "harga": 25000, "kategori": "Makanan" , "pedas_level": 3, "pasangan_menu": "Caramel Macchiato"},
    {"nama_menu": "Cireng", "harga": 20000, "kategori": "Makanan" , "pedas_level": 1, "pasangan_menu": "Cappuccino"},
    {"nama_menu": "Ayam Geprek", "harga": 30000, "kategori": "Makanan" , "pedas_level": 3, "pasangan_menu": "Americano"},
    {"nama_menu": "Tahu Jeletot", "harga": 20000, "kategori": "Makanan" , "pedas_level": 2, "pasangan_menu": "Kopi Dolce"},
    {"nama_menu": "Mie Jebew", "harga": 25000, "kategori": "Makanan" , "pedas_level": 3, "pasangan_menu": "Latte"},
    {"nama_menu": "Bakso Aci", "harga": 25000, "kategori": "Makanan" , "pedas_level": 2, "pasangan_menu": "Avocado Coffee"},
])

print("Jumlah dokumen di collection 'menu':", koleksi_menu.count_documents({}))
print("Daftar database yang ada:", client.list_database_names())
```

**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :
1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.
2. Seluruh menu dengan `harga` di antara 15000 dan 30000.
3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
# (1) Menampilkan seluruh menu kategori Kopi dengan tingkat_kepahitan di atas 3
print("(1) Seluruh menu kategori dengan tingkat_kepahitan di atas 3")
for m in koleksi_menu.find({"tingkat_kepahitan": {"$gt": 3}}):
    print(f"  - {m['nama_menu']} > tingkat_kepahitan: {m['tingkat_kepahitan']}")
print()
    
# (2) Menampilkan seluruh menu dengan harga di antara 15000 dan 30000
print("(2) Seluruh menu dengan harga di antara 15000 dan 30000")
for m in koleksi_menu.find({"harga": {"$gte": 15000, "$lte": 30000}}):
    print(f"  - {m['nama_menu']} > harga: {m['harga']}")
print()

# (3) Menu yang tidak memiliki field pedas_level sama sekali
print("(3) Menu yang tidak memiliki field pedas_level sama sekali")
for m in koleksi_menu.find({"pedas_level": {"$exists": False}}):
    print(f"  - {m['nama_menu']} > harga: Rp{m['harga']:,}, kategori: {m['kategori']}")
```

**C. Pembaruan Data (Update)** *(bobot 15%)*

Tambahkan field `status` bernilai `"Tersedia"` ke seluruh dokumen menggunakan `update_many()`. Kemudian, ubah `status` menjadi `"Habis"` khusus untuk 2 menu pilihan anda menggunakan `update_one()`.

```python
# (1) Menambahkan field status bernilai Tersedia ke seluruh dokumen
koleksi_menu.update_many({}, {"$set": {"status": "Tersedia"}})
print("Setelah update:")
for m in koleksi_menu.find({}, {"_id": 0, "nama_menu": 1, "harga": 1, "kategori": 1, "status": 1}):
    print(f"  - {m['nama_menu']} > harga: Rp{m['harga']:,}, kategori: {m['kategori']}, status: {m['status']}")
print()

# (2) Mengubah status menjadi bernilai Habis pada 2 menu pilihan yang di-update
koleksi_menu.update_one({"nama_menu": "Americano"}, {"$set": {"status": "Habis"}})
koleksi_menu.update_one({"nama_menu": "Ayam Geprek"}, {"$set": {"status": "Habis"}})
print("Setelah mengubah status:")
for m in koleksi_menu.find({}, {"_id": 0, "nama_menu": 1, "harga": 1, "kategori": 1, "status": 1}):
    print(f"  - {m['nama_menu']} > harga: Rp{m['harga']:,}, kategori: {m['kategori']}, status: {m['status']}")
```

**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.
2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg
import pandas as pd

spark = SparkSession.builder \
    .appName("Pertemuan9-MongoDBSpark") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

# Mengubah menjadi Spark DataFrame
dokumen_menu = list(koleksi_menu.find({}))
df_pandas = pd.DataFrame(dokumen_menu)
df_pandas["_id"] = df_pandas["_id"].astype(str)
df_pandas = df_pandas.astype({"tingkat_kepahitan": "Int64", "pedas_level": "Int64"})
df_pandas = df_pandas.astype(object).where(pd.notna(df_pandas), None)

# (1) Menampilkan hasil konversi 
df_spark_menu = spark.createDataFrame(df_pandas)
df_spark_menu.printSchema()
df_spark_menu.show()

# (2) Menampilkan rata-rata harga per kategori
df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
).show()

spark.stop()
print("SparkSession ditutup.")
```

**E. Kesimpulan** *(bobot 20%)*

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?

**Kesimpulan:** Tantangan yang ditemui saat mengonversi data MongoDB menjadi Spark DataFrame adalah mengatur terlebih dahulu field yang tidak seragam antar dokumen dan menyesuaikan `_id` bertipe ObjectId agar dikenali oleh Spark. Seluruh menu dengan kategori `Kopi` tidak memiliki field `pedas_level` dan `pasangan_menu`, sedangkan menu dengan kategori `Makanan` tidak memiliki field `tingkat_kepahitan`. Saat data diubah pada Spark DataFrame yang mengharuskan skema tabel seragam, field yang tidak dimiliki pada kategori tertentu akan bernilai NaN. Akibatnya kolom angka seperti `tingkat_kepahitan` dan `pedas_level` akan berubah menjadi float jika tidak ditangani. Kolom teks seperti field `pasangan_menu` tetap aman karena terbaca sebagai string. 

Cara mengatasinya yaitu dengan mengubah `_id` menjadi bertipe string, mengubah kolom angka ke tipe Int64 agar tetap menjadi bilangan bulat dan menampung nilai kosong, lalu seluruh DataFrame diubah ke tipe object dan mengganti nilai NaN dengan None agar field kosong terbaca sebagai null di Spark. Dengan cara ini, DataFrame dapat dibuat tanpa error, kolom angka tetap berupa bilangan bulat meskipun bertipe long, dan field yang kosong terbaca sebagai null, bukan angka palsu.
